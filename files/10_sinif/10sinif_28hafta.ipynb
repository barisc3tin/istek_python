## 1. Isınma: Dictionary nedir?

Dictionary, verileri **anahtar (key) - değer (value)** ilişkisi ile saklar.

Örnek:
- `"ad"` → `"Ali"`
- `"not"` → `85`

Aşağıdaki hücreyi çalıştırınız.

```python
ogrenci = {
    "ad": "Ali",
    "sinif": "10A",
    "not": 85
}

print(ogrenci)
print("Öğrenci adı:", ogrenci["ad"])
```

## 9. Yuvalı dictionary

Bir dictionary içinde başka dictionary'ler tutabiliriz.

Bu yapı, birden fazla öğrenciyi tek sistem içinde saklamak için çok uygundur.

```python
ogrenci_sistemi = {
    "Ali": {
        "sinif": "10A",
        "matematik": 80,
        "fizik": 75
    },
    "Ayşe": {
        "sinif": "10B",
        "matematik": 95,
        "fizik": 88
    }
}

print(ogrenci_sistemi)
```

## 10. TODO: Yeni öğrenci ekle

Aşağıdaki `ogrenci_sistemi` yapısına yeni bir öğrenci ekleyiniz.

```python
# TODO: Yeni bir öğrenci ekleyiniz
ogrenci_sistemi["Zeynep"] = {
    "sinif": "10A",
    "matematik": 91,
    "fizik": 89
}

print(ogrenci_sistemi)
```

## 11. TODO: Bir öğrencinin notunu güncelle

Aşağıdaki hücrede bir öğrencinin not bilgisini değiştiriniz.

```python
# TODO: Bir öğrencinin notunu güncelle
ogrenci_sistemi["Ali"]["matematik"] = 85

print(ogrenci_sistemi["Ali"])
```

## 12. Dosyaya yazma

Şimdi verilerimizi bir dosyaya kaydedeceğiz.

Bu örnekte veriyi `.json` formatında saklıyoruz. Çünkü dictionary yapısı için çok uygundur.

```python
import json

with open("ogrenci_bilgileri.json", "w", encoding="utf-8") as dosya:
    json.dump(ogrenci_sistemi, dosya, ensure_ascii=False, indent=4)

print("Dosya kaydedildi.")
```

## 13. Dosyadan okuma

Kaydettiğimiz dosyayı tekrar okuyabiliriz.

```python
with open("ogrenci_bilgileri.json", "r", encoding="utf-8") as dosya:
    kayitli_veri = json.load(dosya)

print(kayitli_veri)
```

## 14. TODO: Dosyadan okunan veriyi güncelle

Aşağıdaki hücrede dosyadan okunan veri üzerinde değişiklik yapınız.

```python
# TODO: Bir öğrencinin notunu veya sınıfını güncelleyiniz
kayitli_veri["Ayşe"]["fizik"] = 90

print(kayitli_veri["Ayşe"])
```

## 15. Güncellenmiş veriyi tekrar dosyaya yazma

```python
with open("ogrenci_bilgileri_guncel.json", "w", encoding="utf-8") as dosya:
    json.dump(kayitli_veri, dosya, ensure_ascii=False, indent=4)

print("Güncellenmiş dosya kaydedildi.")
```
